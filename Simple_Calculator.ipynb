```python
def add (a,b):
  return a + b

def sub (a,b):
  return a-b

def mul (a,b):
  return a*b

def div (a,b):
  return a/b

print("Select an Option: ")
print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")

choice = int(input("Enter your Choice between 1 to 4: "))

# Check if choice is valid before asking for numbers
if choice < 1 or choice > 4:
   print('Invalid Choice')
else:
   num1 = float(input("Enter First Number: "))
   num2 = float(input("Enter Second Number: "))

   if choice == 1:
      print('Addition of {0} and {1} is {2}'.format(num1,num2,add(num1,num2)))
   elif choice == 2:
          print('Subtraction of {0} and {1} is {2}'.format(num1,num2,sub(num1,num2)))
   elif choice == 3:
          print('Multiplication of {0} and {1} is {2}'.format(num1,num2,mul(num1,num2)))
   else : # choice must be 4 here
          print('Division of {0} and {1} is {2}'.format(num1,num2,div(num1,num2)))
```
